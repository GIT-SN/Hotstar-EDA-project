# 📊 EDA Insights — Categorical Variables

## 1. Overview of the Dataset

- The dataset contains **33,171 records** and **23 columns**.
- Each record represents an episode-level observation, with information about the platform, genre, episode characteristics, viewer behavior, and retention-related metrics.
- The dataset combines content characteristics such as `pacing_score`, `hook_strength`, and `visual_intensity` with viewer behavior metrics such as `avg_watch_percentage`, `pause_count`, `rewind_count`, and `drop_off_probability`.
- Categorical variables include platform, genre, dialogue density, attention required, retention risk, and other binary indicators.

# 2. Platform Distribution

A hexbin plot was used to visualize the relationship between average watch

## 9. Drop-Off Analysis

The `drop_off` column represents the recorded binary outcome, whereas `drop_off_probability` represents the associated estimated probability. The mean drop-off probability is approximately 0.478, with a median of 0.471.

### 9.3 Numerical Factors Associated with Drop-Off

Comparing numerical variables between the two outcome groups reveals several notable differences:

### 9.4 Key Takeaway

Lower average watch percentage, lower pacing and hook-strength scores, more pauses, and higher cognitive-load scores are associated with actual drop-off records in this dataset. However, potential dependencies between the outcome and its associated variables must be investigated before using these patterns for predictive modelling or business recommendations.

### EDA Concept: Data Leakage and Derived Variables

A derived variable is a column created using another variable or a predefined rule.

In this dataset, `drop_off` matches the rule:

`drop_off = 1` when `drop_off_probability >= 0.60`, otherwise `drop_off = 0`.

The rule produces zero mismatches across all 33,171 records.

This suggests that `drop_off` may have been generated using a threshold on `drop_off_probability`. The exact data-generation process should be verified using the dataset documentation or source code.

**Data leakage:** If a machine-learning model uses a variable that directly determines its target, the model can appear highly accurate without learning meaningful patterns.

Before model training, investigate how the target was generated and exclude variables that directly encode the target or would not be available at prediction time.

- **Pacing Score:** The drop-off group has a lower mean pacing score (4.16) than the non-drop-off group (5.61).

### 4. Important Data-Quality Finding: Drop-Off Threshold

A threshold check showed that the actual `drop_off` label matches the following rule across all 33,171 records:

- `drop_off = 1` when `drop_off_probability >= 0.60`
- `drop_off = 0` when `drop_off_probability < 0.60`

The rule produces **zero mismatches**.

Furthermore, `retention_risk` aligns perfectly with the observed drop-off labels: all 4,804 drop-off records are labelled high risk, while all records without drop-off are labelled low or medium risk.

This strongly suggests that these columns are deterministically linked. However, the exact process used to generate the dataset should be verified before concluding how the labels were created.

### Conclusion

The observed data associates higher drop-off with lower watch percentage, lower pacing scores, lower hook strength, more pauses, and higher cognitive load. Skip-intro status, dialogue density, genre, and other categorical features also show differences in drop-off rates.

However, the perfect threshold relationship between `drop_off_probability` and `drop_off`, along with the perfect alignment of `retention_risk`, raises a potential data-leakage concern. These relationships must be investigated before using the columns for predictive modelling. The findings should be treated as descriptive associations rather than causal explanations.

7. **Content-level differences:** Some titles have higher median watch percentages than others, although the rankings use relatively small samples.
8. **Data quality:** An extreme episode-duration value of 1,225 minutes was identified and retained in the original dataset, with the extreme value excluded only from the relevant visualisation.
9. **Target relationship:** The actual drop-off label matches a threshold of 0.60 on drop-off probability. This relationship must be considered when interpreting drop-off findings.

## Business Recommendations

- Investigate which content characteristics are associated with higher engagement, particularly pacing and hook strength.
- Explore why observed watch percentages differ across genres without assuming that genre alone explains the differences.
- Review high-pause and high-drop-off content to identify possible viewer-experience issues.
- Use title-level rankings as a starting point for further investigation, not as the sole basis for content investment decisions.
- Validate suspicious target relationships and data-quality anomalies before using the dataset for predictive modelling.
- Collect reliable, time-stamped viewing behaviour if future analysis needs to predict drop-off while an episode is playing.

## Limitations

- The findings are observational and do not establish causation.
- Sample sizes differ across genres, platforms, and titles.
- Episode numbers are not unique across shows, and season number has no variation.
- Some variables may be derived from the same underlying rules, so their relationships may not represent independent evidence.
- The dataset's generation logic and the representativeness of its records should be verified before making real-world business decisions.
